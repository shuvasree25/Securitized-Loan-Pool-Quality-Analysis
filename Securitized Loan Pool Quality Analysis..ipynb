Converting Data into string, float

```python
df["loan_id"] = df["loan_id"].astype(str)
df["customer_id"] = df["customer_id"].astype(str)
df["customer_name"] = df["customer_name"].astype(str)
df["gender"] = df["gender"].astype(str)
df["country"] = df["country"].astype(str)
df["state"] = df["state"].astype(str)

df["age"] = df["age"].astype(int)
df["employment_type"] = df["employment_type"].astype(str)
df["loan_term"] = df["loan_term"].astype(int)
df["loan_type"] = df["loan_type"].astype(str)
```

```python
df["origination_date"] = pd.to_datetime(df["origination_date"], errors="coerce")
```

Convert the data into integer and fill the missing value

```python
df["income"] = pd.to_numeric(df["income"], errors="coerce")
df["income"] = df["income"].fillna(0)
df["income"] = df["income"].astype("int64")

df["credit_score"] = pd.to_numeric(df["credit_score"], errors="coerce")
df["credit_score"] = df["credit_score"].fillna(0)
df["credit_score"] = df["credit_score"].astype("int64")

df["interest_rate"] = pd.to_numeric(df["interest_rate"], errors="coerce")
df["interest_rate"] = df["interest_rate"].fillna(0)
df["interest_rate"] = df["interest_rate"].astype("float64")
```

```python
df.drop(df.columns[-1], axis=1, inplace=True)
```
